 Examination performance analysis: Generate a reproducible 12 x 4 NumPy array of integer test marks from 20 to 100, inclusive, using a fixed seed. Calculate each student's average and each test's mean and standard deviation. Count students passing all four tests (at least 40 in each). Identify students with averages of at least 75 and the test with greatest variation. State your student numbering convention

```python
arr = np.random.randint(20, 101, size=(12, 4))
print(arr)
```

```python
# Generate reproducible marks (12 students, 4 tests)
np.random.seed(42)
marks = np.random.randint(20, 101, size=(12, 4))

print("Marks of 12 Students:")
print(marks)

# Calculate each student's average (row-wise)
student_avg = np.mean(marks, axis=1)
print("\nAverage Marks of Each Student:")
print(np.round(student_avg, 2))

# Calculate each test's mean (column-wise)
test_mean = np.mean(marks, axis=0)
print("\nMean of Each Test:")
print(np.round(test_mean, 2))

# Calculate each test's standard deviation
test_std = np.std(marks, axis=0)
print("\nStandard Deviation of Each Test:")
print(np.round(test_std, 2))

# Count students passing all four tests (at least 40 in each)
pass_all = np.all(marks >= 40, axis=1)
print("\nNumber of Students Passing All Four Tests:")
print(np.sum(pass_all))

# Identify students with averages >= 75
high_performers = np.where(student_avg >= 75)[0] + 1
print("\nStudents with Average >= 75 (Student Numbers):")
print(high_performers)

# Identify the test with greatest variation
most_variable_test = np.argmax(test_std) + 1
print("\nTest with Greatest Variation:")
print("Test", most_variable_test)

```

 Matrix transformation and analysis: Construct a 5 x 5 matrix containing integers 21 to 45. Extract the central 3 x 3 matrix, transpose it, and calculate the matrix product of the extracted matrix with its transpose. Calculate row-wise sums of the product and identify its maximum and all positions tied at that maximum. Explain dimensions at every stage.

```python
# Step 1: Create a 5 x 5 matrix containing integers 21 to 45
A = np.arange(21, 46).reshape(5, 5)
print("Original Matrix (5 x 5):")
print(A)

# Step 2: Extract the central 3 x 3 matrix
B = A[1:4, 1:4]
print("\nCentral Matrix (3 x 3):")
print(B)

# Step 3: Transpose the extracted matrix
BT = B.T
print("\nTranspose of Central Matrix (3 x 3):")
print(BT)

# Step 4: Calculate matrix product
C = B @ BT
print("\nMatrix Product (3 x 3):")
print(C)

# Step 5: Calculate row-wise sums
row_sums = np.sum(C, axis=1)
print("\nRow-wise Sums:")
print(row_sums)

# Step 6: Find maximum row sum
maximum = np.max(row_sums)
print("\nMaximum Row Sum:")
print(maximum)

# Step 7: Find all positions tied at the maximum
positions = np.where(row_sums == maximum)[0]
print("\nPositions of Maximum Row Sum (0-based):")
print(positions)

print("\nPositions of Maximum Row Sum (1-based):")
print(positions + 1)

```

 Sales and revenue analysis: Create a 6 x 4 matrix of quantities sold over six days for four products. Use unit prices [150, 240, 90, 320]. Calculate daily revenue, total revenue per product, the highest-revenue day, and the best-selling product by quantity. Reduce prices by 10% for products whose total quantity exceeds 100, then recalculate total revenue without modifying original prices.

```python
sales = np.array([
    [20, 15, 30, 10],
    [25, 20, 25, 15],
    [18, 25, 20, 12],
    [30, 18, 35, 20],
    [22, 30, 28, 18],
    [28, 22, 32, 25]
])


prices = np.array([150, 240, 90, 320])


print(sales)
print(prices)


daily_revenue = np.sum(sales * prices, axis=1)
print("\nDaily Revenue:")
print(daily_revenue)


product_revenue = np.sum(sales * prices, axis=0)
print("\nTotal Revenue per Product:")
print(product_revenue)


highest_day = np.argmax(daily_revenue) + 1
print("\nHighest Revenue Day:")
print("Day", highest_day)


product_quantity = np.sum(sales, axis=0)
best_product = np.argmax(product_quantity) + 1

print("\nTotal Quantity Sold per Product:")
print(product_quantity)

print("\nBest-Selling Product:")
print("Product", best_product)

# Step 6: Reduce prices by 10% for products with total quantity > 100
discounted_prices = np.where(
    product_quantity > 100,
    prices * 0.90,
    prices
)

print("\nDiscounted Unit Prices:")
print(discounted_prices)


discounted_revenue = np.sum(sales * discounted_prices)
print(np.sum(product_revenue))
print(discounted_revenue)
print(prices)

```

 Random data cleaning and statistics: Generate 100 normal observations with mean 50 and standard deviation 12 using a fixed seed. Count values outside 30-70 and replace them with the nearest boundary using np.where(). Compare original and cleaned mean, variance, standard deviation, minimum, and maximum. Explain the statistical effects of boundary replacement.

```python
np.random.seed(42)
data = np.random.normal(50, 12, 100)

# Count values outside 30–70
count = np.sum((data < 30) | (data > 70))

# Replace out-of-range values with nearest boundary
cleaned = np.where(data < 30, 30, np.where(data > 70, 70, data))

print("Values outside range:", count)

print("\nOriginal Data:")
print("Mean:", np.mean(data))
print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))

print("\nCleaned Data:")
print("Mean:", np.mean(cleaned))
print("Variance:", np.var(cleaned))
print("Standard Deviation:", np.std(cleaned))
print("Minimum:", np.min(cleaned))
```

```python
branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})

# Ordinary addition
combined = branch_a + branch_b + branch_c

print(combined)

# Quarters common to all branches
common = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)

print(common)


print(combined.loc[common])

# Treat missing branch records as zero
combined_zero = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)

print(combined_zero)
```
