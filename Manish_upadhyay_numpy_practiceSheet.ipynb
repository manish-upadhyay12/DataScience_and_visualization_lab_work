**2. Array Arithmetic**

```python
#6.	Given two NumPy arrays, perform element-wise addition, subtraction, multiplication, and division2 =
arr1 = np.array([1,2,3,4])
arr2 = np.array([2,5,6,7])
print(arr1 + arr2)
print(arr1-arr2)
print(arr1*arr2)
print(arr1/arr2)
```

```python
# 7.	Calculate the square and cube of every element in a NumPy array.
arr = np.array([1,2,3,4])
print(arr**2)
print(arr**3)
```

```python
#8.	Increase every element of an array by 15%.
arr = np.array([1,2,3,4])
print(arr*1.15)
```

```python
#9.	Given two arrays, find their element-wise maximum and minimum.
arr1 = np.array([1,2,3,4])
arr2= np.array([5,6,7,8])
print(np.maximum(arr1,arr2))
print(np.minimum(arr1,arr2))
```


**3. Indexing & Slicing**

```python
# 10.	Create an array from 0 to 20 and extract the first five elements, last five elements, elements from index 5 to 15, and every second element.

arr  = np.arange(0,21)
print(arr[:5])
print(arr[-5:])
print(arr[5:16])
print(arr[::2])
```

```python
# 11.	Replace the first five elements of an array with 100.
arr = np.arange(0,21)
arr[:5]  = 100
arr[:5]
```

```python
# 12.	Create an array and demonstrate the difference between a slice/view and a copy


# Create an array
arr = np.array([10, 20, 30, 40, 50])

# Slice / View
view = arr[1:4]

# Copy
copy = arr[1:4].copy()

# Change the view
view[0] = 100

# Change the copy
copy[1] = 200

print("Original array:", arr)
print("View:", view)
print("Copy:", copy)
```
