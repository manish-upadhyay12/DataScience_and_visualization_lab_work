29. Write a short Pandas program that creates a labelled Series for four products and sales, displays its values
and index, selects one label, filters sales above 500, checks whether a label exists, and converts the Series to a
dictionary.

```python

# Create a labelled Series
sales = pd.Series(
    [400, 650, 800, 300],
    index=['Laptop', 'Mobile', 'Tablet', 'Keyboard']
)

# Display values and index
print("Values:", sales.values)
print("Index:", sales.index)

# Select one label
print("Mobile Sales:", sales['Mobile'])

# Filter sales above 500
print("Sales above 500:\n", sales[sales > 500])

# Check whether a label exists
print("Laptop exists:", 'Laptop' in sales.index)

# Convert Series to dictionary
print("Dictionary:", sales.to_dict())

```
